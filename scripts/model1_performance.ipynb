```python
import joblib
import pandas as pd
from sklearn.metrics import (
    accuracy_score,
    f1_score,
    roc_auc_score,
    confusion_matrix,
    classification_report
)

# Then load the saved version of the Logistic rgeression model 
logreg =joblib.load("../models/logreg_model.pkl")

# Loading the processed data split 
X_train, X_test, y_train, y_test = joblib.load("../datasets/train_test_split.pkl")

print("Logistic Regression model and test dataset loaded successfully!")
print("X_test shape:", X_test.shape)
```

Show predictions and evaluate performance

```python
# Start with predicting class labels and class probabilities on the test set
y_pred = logreg.predict(X_test)
y_pred_proba = logreg.predict_proba(X_test)[:,1]

# Change labels to number by saying no=0 and yes=1
y_test_num = y_test.map({"no": 0, "yes": 1})
y_pred_num =pd.Series(y_pred).map({"no": 0, "yes": 1})

# Show the main metrics
print("=== Logistic Regression Perfroamnce Metrics ===")
print("Accuracy:", accuracy_score(y_test_num, y_pred_num))
print("F1 score (yes):", f1_score(y_test_num, y_pred_num, pos_label=1))
print("ROC AUC score:", roc_auc_score(y_test_num, y_pred_proba))

# Show the evlaution 
print("\nConfusion Matrix:\n", confusion_matrix(y_test_num, y_pred_num))
print("\nClassifcation Report:\n", classification_report(y_test_num, y_pred_num))
```
