Get the saved test data and saved Random Forest model


```python
import joblib
import pandas as pd

# loading the saved test split and trained data Random Forest
_, X_test, _, y_test = joblib.load("../datasets/train_test_split.pkl")
rf = joblib.load("../models/rf_model.pkl")

print("Random Forest model and test dataset loaded successfully!")
print("X_test shape:", X_test.shape)
```

Show predictions and performance metrics

```python
from sklearn.metrics import (
    accuracy_score,
    f1_score,
    roc_auc_score,
    confusion_matrix,
    classification_report
)

# prediction using the loaded model
y_pred_rf =rf.predict(X_test)
y_pred_proba_rf = rf.predict_proba(X_test)[:, 1]

# change labels to numbers
y_test_num = y_test.map({"no": 0, "yes": 1})
y_pred_num = pd.Series(y_pred_rf).map({"no": 0, "yes": 1})

# show results
print("===Random Forest Metrics ===")
print("Accuracy:", accuracy_score(y_test_num, y_pred_num))
print("F1 score (yes):", f1_score(y_test_num, y_pred_num, pos_label=1))
print("ROC AUC score:", roc_auc_score(y_test_num, y_pred_proba_rf))

print("\nConfusion Matrix:\n", confusion_matrix(y_test_num, y_pred_num))
print("\nClassification Report:\n", classification_report(y_test_num, y_pred_num))
```
