```python
# import pandas as pd
import pandas as pd
import numpy as np
```

Loading my data set


```python
df = pd.read_csv(
    "../datasets/bank-additional-full.csv",
    sep=';'
)
```

```python
# Inspect the data set
df.shape  # check the rows and columns the data set has
df.columns  # inspect the column names
df.dtypes  # data types there is
df.head()  # show the first few (5) rows
```

Find if there are any missing values, usually shown as unknown

```python
# Find if there are any missing values, usually shown as unknown
# by counting the "unknown" per column
unknown_counts = (df =="unknown").sum()
print(unknown_counts[unknown_counts > 0])
```

Find if Rows have been duplicated

```python
duplicates = df.duplicated().sum()
print("Duplicate rows:", duplicates)
```

```python
# Since found 12, drop the duplicates
df = df.drop_duplicates()

# Confirm after dropping
print("Shape after dropping duplicates:", df.shape)
```

Drop the duration column for leakage

```python
df = df.drop(columns=["duration"]) # dropping the duration column because call length is only known after contact, avoiding leaking information into the model
```

Check targeted variable y for class balance in where "yes" subscribed or "no"

```python
df["y"].value_counts()
df["y"].value_counts(normalize=True)
```

Save the cleaned version of the dataset
