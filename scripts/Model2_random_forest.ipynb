Loading the processed train test data

```python
import joblib
import pandas as pd

# Making sure to laod the exact train test split from the feature enigineering
X_train, X_test, y_train, y_test = joblib.load("../datasets/train_test_split.pkl")

print("Data loaded successfully!")
print("X_train shape:", X_train.shape)
print("X_test shape:", X_test.shape)
```

Starting the Random Forest Classifer training

```python
from sklearn.ensemble import RandomForestClassifier

# The random forest model hyperparameters will be:

rf = RandomForestClassifier(
    n_estimators=100,
    max_depth=15,
    min_samples_split=2,
    min_samples_leaf=1,
    max_features="sqrt",
    class_weight="balanced",
    random_state=42,
    n_jobs=-1
    
)

# Then train the model
rf.fit(X_train, y_train)

print("Random Forest model trained successfully!")
```

Have the predictions and evaluation

```python
from sklearn.metrics import (
    accuracy_score,
    f1_score,
    roc_auc_score,
    confusion_matrix,
    classification_report
)

# show prediction class labels and probabilities based on the test set
y_pred_rf =rf.predict(X_test)
y_pred_proba_rf = rf.predict_proba(X_test)[:,1]

# Conver the labels to numbers where no=0 and yes =1 
y_test_num = y_test.map({"no":0, "yes":1})
y_pred_num = pd.Series(y_pred_rf).map({"no": 0, "yes": 1})

# show the main performance metrics
print("Accuracy:", accuracy_score(y_test_num, y_pred_num))
print("F1 score (yes):", f1_score(y_test_num, y_pred_num, pos_label=1))
print("ROC AUC score:", roc_auc_score(y_test_num, y_pred_proba_rf))

# show the more detailed evaluation outputs
print("\nConfusion Matrix:\n", confusion_matrix(y_test_num, y_pred_num))
print("\nClassification Report:\n", classification_report(y_test_num, y_pred_num))

```

Saving my trained model

```python
import joblib

#Saving the trained Random Forset model
joblib.dump(rf, "../models/rf_model.pkl")

print("Random Forest model successfully saved to '../models/rf_model.pkl'")
```
